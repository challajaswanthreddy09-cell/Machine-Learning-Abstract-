## 3. Data Leakage Prevention (Crucial ML Principle)
A valid credit scoring model must ONLY use data available at loan origination.  
The following columns are post-origination information and cause severe target leakage:
- `recoveries`: 0 for all Good Loans; positive only after charge-off.
- `total_pymnt` & `total_rec_prncp`: Accrued payments received over the lifetime of the loan.
- `final_d`: Date of final settlement/default.
- `issue_d` & `year`: Administrative origination timestamps.
- `id`: Arbitrary row identifier.

```python
print('Evidence of Target Leakage in recoveries:')
print(df.groupby('loan_condition')['recoveries'].agg(['count', 'mean', 'max']))

# Drop leakage columns
project_root = os.path.abspath(os.path.join('..', '..'))
sys.path.append(project_root)
from ml.preprocessing import load_and_clean_data, NUMERICAL_FEATURES, CATEGORICAL_FEATURES
X, y = load_and_clean_data(data_path)
print(f'Clean features selected ({len(X.columns)}):', list(X.columns))
```
